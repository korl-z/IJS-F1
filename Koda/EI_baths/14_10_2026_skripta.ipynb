```python
from ei_ness import Grid, Band, MF, Bath, EI, diag

gr = Grid(4, 4, vol=16.0)
bd, mf = Band(g=2.0, ta=1.0, tb=-1.0), MF(v=4.4, n=1.0)
bs = (Bath(t=1.38, w0=0.1, c=0.3, eta=0.18, d=0.0),
      Bath(t=0.32, w0=0.6, c=0.3, eta=0.18, d=0.0))

s = EI(gr, bd, mf, bs)
d, m = 0.9, -0.6
e, u = diag(s.e0, mf, d, m)
mu = (e - mf.v / 2.0) / 0.35
st = s.solve(d, m, mu, tol=1e-8, nmax=3000)

print(st.ok, st.d, st.m, st.na + st.nb, st.err)
```
